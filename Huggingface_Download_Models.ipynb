# Download the model and tokenizer to a local directory

It will cache the files in the local `~/.cache/huggingface/hub` directory first

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "unsloth/Meta-Llama-3.1-8B-Instruct"
model_local_path = "./models"

# Download and save both tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
tokenizer.save_pretrained(model_local_path)
model.save_pretrained(model_local_path)
```

# Download and cache the files in a standard location

```python
import os
from huggingface_hub import snapshot_download

os.environ["HF_HOME"] = "basemodel_tmp"
snapshot_download(repo_id="unsloth/Meta-Llama-3.1-8B-Instruct", repo_type="model")
```

# Check for `bfloat16` support

```python
import torch
from cpuinfo import get_cpu_info

# Check PyTorch version and bfloat16 availability
print(f"PyTorch version: {torch.__version__}")
if hasattr(torch, 'bfloat16'):
    print("torch.bfloat16 is supported by PyTorch")
else:
    print("torch.bfloat16 is NOT supported by PyTorch")
    exit()

# Check GPU support
if torch.cuda.is_available():
    print(f"CUDA available, GPU: {torch.cuda.get_device_name(0)}")
    if torch.cuda.is_bf16_supported():
        print("GPU supports bfloat16")
    else:
        print("GPU does NOT support bfloat16")
else:
    print("No CUDA GPU available")

# Check CPU support
try:
    tensor = torch.tensor([1.0, 2.0], dtype=torch.bfloat16, device='cpu')
    print("bfloat16 is supported on CPU (software):", tensor.dtype)
except Exception as e:
    print("bfloat16 is NOT supported on CPU:", str(e))

# Check CPU hardware acceleration
cpu_info = get_cpu_info()
print("CPU:", cpu_info['brand_raw'])
flags = cpu_info.get('flags', [])
if 'avx512_bf16' in flags or 'amx_bf16' in flags:
    print("CPU supports bfloat16 hardware acceleration")
else:
    print("CPU does NOT support bfloat16 hardware acceleration")
```
